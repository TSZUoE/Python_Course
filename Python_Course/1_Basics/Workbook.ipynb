```python
my_skills = ['Python', 'SQL', 'Excel']
```

```python
job_roles = [
    {'role': 'Data Analyst', 'skills': ['Python', 'SQL', 'Excel']},
    {'role': 'Data Scientist', 'skills': ['Python', 'R', 'Machine Learning', 'Deep Learning']},
    {'role': 'Machine Learning Engineer', 'skills': ['Python', 'TensorFlow', 'PyTorch', 'Scikit-Learn']},
    {'role': 'Data Engineer', 'skills': ['Python', 'Apache Spark', 'Hadoop', 'SQL']},
    {'role': 'Business Intelligence Analyst', 'skills': ['Python', 'SQL', 'Tableau', 'Power BI', 'Excel']},
    {'role': 'Quantitative Analyst', 'skills': ['R', 'Python', 'MATLAB', 'Statistics']},
    {'role': 'Operations Analyst', 'skills': ['Python', 'SQL', 'Data Visualization', 'Process Improvement']},
    {'role': 'Database Administrator', 'skills': ['SQL', 'Oracle', 'MySQL', 'Database Management']},
    {'role': 'AI Engineer', 'skills': ['Python', 'TensorFlow', 'PyTorch', 'Computer Vision']},
    {'role': 'Statistician', 'skills': ['R', 'SAS', 'Python', 'Statistical Modeling']}
]
```

```python
qualified_roles = [] #define an empty list to write into

for job in job_roles: #begin for loop
  qualified = True #assume you are qualified for each role

  for skill in my_skills: #iterate through my skills
    if skill not in job['skills']: #if any of my skills are absent from the skills section of the job
      qualified = False #change qualified value to false
      break #and break the loop
  if qualified: #if I am qualified (i.e., skills are present in the job description)
    qualified_roles.append(job['role']) #then add it to the list of qualified roles
```

```python
qualified_roles
```

```python
#we can write this more succinctly using the all() function

qualified_roles = []

for job in job_roles:
  #check if all required skills are in my_skills
  if all(skill in job['skills'] for skill in my_skills):
    #add it to the list
    qualified_roles.append(job['role'])

qualified_roles
```

# Classes

```python
#define a list class

class TomsList:
    def __init__(self):
        """Initialize an empty list."""
        self._items = []

    def add(self, item):
        """Add an item to the end of the list."""
        self._items.append(item)

    def __getitem__(self, index):
        """Retrieve an item by index."""
        return self._items[index]

    def __setitem__(self, index, value):
        """Set an item at a specific index."""
        self._items[index] = value

    def __repr__(self):
        """Return a string representation of the list."""
        return str(self._items)

    def __len__(self):
        """Return the length of the list."""
        return len(self._items)
```

```python
#create an instance of this class
my_list = TomsList()
my_list
```

```python
my_list.add("Data Analyst")
my_list.add("Finance Analyst")
my_list
```

```python
len(my_list)
```

classes are useful for bundling up methods and make an object behave in a specific way

An example problem using our job_analyser.py code

```python
class BaseSalary:
  def __init__(self, base_salary, bonus_rate=0.1, symbol='£'): #initialise variables
    self.base_salary = base_salary
    self.bonus_rate = bonus_rate
    self.symbol = symbol
    self.total_salary = base_salary * (1 + bonus_rate)
    self.bonus = self.total_salary - self.base_salary

  def __repr__(self): #define method for how output is formatted
    return f'{self.symbol}{self.base_salary:,}' #:, inside of f string formats numbers to have commas

  def show_salary(self): #show salary method is defined
      return f'{self.symbol}{self.total_salary:,}' #define the return value and format

  def show_bonus(self):
      return f'{self.symbol}{self.bonus:,}'
```

```python
salary = BaseSalary(100_000)
salary.show_bonus()
```
